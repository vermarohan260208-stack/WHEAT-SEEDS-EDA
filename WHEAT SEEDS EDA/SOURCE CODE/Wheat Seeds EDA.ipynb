```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('seeds.csv')

# short column names (same order as the file)
df.columns = ['area', 'perimeter', 'compactness', 'length',
              'width', 'asymmetry', 'groove_length', 'variety']

# 1, 2, 3 -> variety names
df['variety'] = df['variety'].map({1: 'Kama', 2: 'Rosa', 3: 'Canadian'})
```

## Question 1 - How many samples belong to each variety?

```
variety_counts = df['variety'].value_counts()
print(variety_counts)
```

## Question 2 - Mean, median, minimum and maximum of area, perimeter and length (NumPy)

```
for col in ['area', 'perimeter', 'length']:
    arr = np.array(df[col])      # NumPy array
    print('---', col, '---')
    print('Mean   :', round(np.mean(arr), 3))
    print('Median :', round(np.median(arr), 3))
    print('Minimum:', np.min(arr))
    print('Maximum:', np.max(arr))
    print()
```

## Question 3 - Which variety has the highest average area and width?

```
avg = df.groupby('variety')[['area', 'width']].mean()
print(avg.round(3))
print()
print('Highest average area :', avg['area'].idxmax())
print('Highest average width:', avg['width'].idxmax())
```

## Question 4 - Sort by kernel area and show the top 10

```
top10 = df.sort_values(by='area', ascending=False).head(10)
top10
```

## Question 5 - Missing values and duplicates

```
print('Missing values per column:')
print(df.isnull().sum())
print()
print('Duplicate rows:', df.duplicated().sum())
```

## Question 6 - Compare average measurements across the three varieties

```
variety_means = df.groupby('variety').mean(numeric_only=True).round(3)
variety_means
```

## Pandas filtering

```
# kernels with an area above the overall average
large = df[df['area'] > np.mean(np.array(df['area']))]
print('Kernels above average area:', len(large))
print(large['variety'].value_counts())

# Rosa kernels with area greater than 19
rosa_big = df[(df['variety'] == 'Rosa') & (df['area'] > 19)]
print('\nRosa kernels with area > 19:', len(rosa_big))
```

## Chart 1 - Number of samples per variety

```
plt.figure(figsize=(6, 4))
plt.bar(variety_counts.index, variety_counts.values, edgecolor='black')
plt.title('Number of samples per variety')
plt.xlabel('Variety')
plt.ylabel('Count')
plt.show()
```

## Chart 2 - Histogram of kernel area

```
plt.figure(figsize=(6, 4))
plt.hist(df['area'], bins=15, color='#4c72b0', edgecolor='black')
plt.title('Distribution of kernel area')
plt.xlabel('Area')
plt.ylabel('Number of kernels')
plt.show()
```

## Three observations
1. The dataset has 199 samples (slightly fewer than the 210 stated) and the three varieties are almost balanced: Rosa 68, Kama 66 and Canadian 65.

2. Rosa has the largest kernels, with the highest average area (18.37) and width (3.68), and all 10 of the largest kernels are Rosa. Canadian has the smallest kernels (average area 11.88).

3. There are no missing values and no duplicate rows, so no cleaning was needed. The area histogram is not a single hump because it mixes three varieties of different sizes, with a rise around 18-19 where the Rosa kernels cluster.
